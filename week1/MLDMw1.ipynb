````python
"""
Dictionaries are useful for storing information in key-value pairs.
We shall import some libraries for this.
"""
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches


colours = {"green": "#00B828", "yellow": "#FFD900", "purple": "#800FF2",
           "blue": "#0073FF", "orange": "#FF5000", "grey": "#B3B3B3"}


def show_colour(colour):
  """
  This function takes a hex code as the input and outputs its colour.
  """
  fig, ax = plt.subplots(figsize=(2, 1))
  ax.add_patch(mpatches.Rectangle((0, 0), 1, 1, color=colours[colour]))
  ax.set_xlim(0, 1)
  ax.set_ylim(0, 1)
  ax.axis('off')
  plt.show()
  return

print(colours["purple"]) # this prints the colour hex code
show_colour("purple")

````

## Common Python scientific libraries

We shall now cover the basics of NumPy, SciPy, and Matplotlib.

NumPy provides support for multidimensional arrays and many mathematical operations.

SciPy offers numerical tools often used in scientific computing, e.g. optimisation, integration, interpolation, and statistical tools.

Matplotlib is a powerful library for creating scientific plots.

````python
import numpy as np
import scipy as sp
import matplotlib.pyplot as plt
````

### NumPy

````python
"""
Here are some ways to create a NumPy array, i.e. a vector or matrix.
"""

# Explicitly
my_array = np.array([1, 2, 3, 4, 5])
print(my_array)

# From a list
python_list = [1, 2, 3, 4, 5]
my_array = np.array(python_list)
print(my_array)

# Using array generators
my_array = np.arange(1, 6)
print(my_array)

# A bit fancier
print("""
Now a bit fancier.
""")

my_array = np.arange(1, 10.5, 0.5) # in 0.5 increments
print(my_array)
print("")
my_array = np.linspace(0, 10, 51) # make 51 points from 0 to 10 inclusive
print(my_array)
````

Now let's do some linear algebra.



````python
my_matrix = np.array([[2, -1, 0],
                      [1, 2, 1],
                      [0, 1, 3]])

print(f"""
  Matrix is:
  {my_matrix}
  """)

print(f"""
  Determinant is:
  {np.linalg.det(my_matrix):.2f}
  """)

print(f"""
  Transpose is:
  {my_matrix.T}
  """)
````
