````python
"""
Dictionaries are useful for storing information in key-value pairs.
We shall import some libraries for this.
"""
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches

colours = {"green": "#00B828", "yellow": "#FFD900", "purple": "#800FF2",
           "blue": "#0073FF", "orange": "#FF5000", "grey": "#B3B3B3"}

def show_colour(colour):
  """
  This function takes a hex code as the input and outputs its colour.
  """
  fig, ax = plt.subplots(figsize=(2, 1))
  ax.add_patch(mpatches.Rectangle((0, 0), 1, 1, color=colours[colour]))
  ax.set_xlim(0, 1)
  ax.set_ylim(0, 1)
  ax.axis('off')
  plt.show()
  return

print(colours["purple"]) # this prints the colour hex code
show_colour("purple")

````

### Matplotlib

Matplotlib is very versatile: https://matplotlib.org/stable/gallery/index.html. We will just do a few simple plots.

Tip: don't be afraid to use AI assistants to beautify your plots!

````python
import numpy as np
import matplotlib as mpl
import matplotlib.pyplot as plt
colours = {"green": "#00B828", "yellow": "#FFD900", "purple": "#800FF2",
           "blue": "#0073FF", "orange": "#FF5000", "grey": "#B3B3B3"}

# Create some x and y data.
x = np.linspace(0, 2 * np.pi, 100)
y1 = np.sin(x)
y2 = np.cos(x)

# Create a mpl object.
fig, ax = plt.subplots()
ax.set_xlim(0, 2 * np.pi)
ax.set_ylim(1.1 * np.min([y1, y2]), 1.1 * np.max([y1, y2]))
ax.set_xlabel(r"$\theta$")
ax.set_ylabel(r"$\sin(\theta)$")


# Add data into the mpl object.
ax.plot(x, y1, color=colours["blue"], linewidth=3, label=r"$\sin(\theta)$")
ax.scatter(x, y2, color=colours["orange"], label=r"$\cos(\theta)$")
ax.legend()
# plt.savefig("sin_cos.pdf") # uncomment this and click the folder icon
plt.show()
````

### SciPy

Now we will look at some basic functionalities of SciPy.

#### Interpolation

````python
from scipy.interpolate import interp1d

# Create sparse data.
x_sparse = np.array([0, 1, 2, 3, 5, 8])
y_sparse = np.array([1, 3, 2, 5, 4, 7])

# Create interpolation function.
f_cubic = interp1d(x_sparse, y_sparse, kind="cubic")
# Ask Gemini which other interpolation kinds exist!

# Create a dense x grid and evaluate the interpolation function.
x_dense = np.linspace(0, 8, 100)
y_cubic = f_cubic(x_dense)

plt.plot(x_sparse, y_sparse,
         label="sparse", color=colours["blue"], linewidth=3)
plt.plot(x_dense, y_cubic,
         label="interpolated", color=colours["orange"], linewidth=3)
plt.legend()
plt.show()
````

#### Integration

````python
from scipy.integrate import quad

# Define a function.
def gaussian(x):
  return np.exp(-x**2)

result, error = quad(gaussian, -np.inf, np.inf)
print(f"""Approximate result is
{result:.20f}.
""")
print(f"""Exact solution is
{np.sqrt(np.pi):.20f}.
""")
````

#### Linear regression

````python
import numpy as np
from scipy import stats
import matplotlib.pyplot as plt

# First we will generate some data with noise.
x = np.linspace(0, 10, 50)
y = 2.5 * x + 1.5 + np.random.normal(0, 2, 50)

# Do the linear regression.
slope, intercept, r_value, p_value, std_err = stats.linregress(x, y)

print(f"Slope: {slope:.3f}")
print(f"Intercept: {intercept:.3f}")
print(f"R² = {r_value**2:.3f}")
print("")

"""
Now let's plot the data and the regression line.
Note that this is a "quick and dirty" plot, i.e. we do not
create the mpl object first and then add data to it.
"""
plt.scatter(x, y, alpha=0.5, label='Data')
plt.plot(x, slope*x + intercept, 'r-',
         label=f"fit: $y={slope:.2f}x+{intercept:.2f}$")
plt.legend()
plt.show()
````
